# Retail Data Platform - üzleti elemzés

Ez a notebook közvetlenül a PostgreSQL szolgáltatáshoz kapcsolódik, és bemutatja a legfontosabb üzleti mutatókat.

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
from sqlalchemy import create_engine

url = (
    f"postgresql+psycopg://{os.environ['DB_USER']}:{os.environ['DB_PASSWORD']}"
    f"@{os.environ['DB_HOST']}:{os.environ['DB_PORT']}/{os.environ['DB_NAME']}"
)
engine = create_engine(url)
print('Kapcsolat létrejött.')
```

```python
query = '''
SELECT region, category, COUNT(*) AS orders, SUM(quantity) AS units,
       ROUND(SUM(revenue), 2) AS revenue
FROM orders
GROUP BY region, category
ORDER BY revenue DESC
'''
summary = pd.read_sql(query, engine)
summary.head(10)
```

```python
regional = summary.groupby('region')['revenue'].sum().sort_values()
regional.plot(kind='barh', title='Bevétel régiónként', color='#5794F2')
plt.xlabel('Bevétel (USD)')
plt.tight_layout();
```
